```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Step 1: Load the dataset
data = pd.read_csv('/content/Banknote-authentication-dataset-.csv')

# Display basic information about the dataset
print("Dataset Overview:")
print(data.info())
print("\nFirst 5 rows of the dataset:")
print(data.head())

# Step 2: Calculate statistical measures
print("\nStatistical Summary:")
print(data.describe())

# Step 3: Visualize the data
# Scatter plot of V1 vs V2
plt.figure(figsize=(8, 6))
sns.scatterplot(x='V1', y='V2', data=data, s=50, alpha=0.7)
plt.title('Scatter Plot of V1 vs V2')
plt.xlabel('V1')
plt.ylabel('V2')
plt.grid(True)
plt.show()

# Histograms for V1 and V2
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
sns.histplot(data['V1'], kde=True, bins=30, color='blue')
plt.title('Histogram of V1')
plt.xlabel('V1')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
sns.histplot(data['V2'], kde=True, bins=30, color='green')
plt.title('Histogram of V2')
plt.xlabel('V2')
plt.ylabel('Frequency')

plt.tight_layout()
plt.savefig('histograms_v1_v2.png')  # Save histograms
plt.show()

# Pairwise correlation heatmap
correlation_matrix = data.corr()
plt.figure(figsize=(6, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()

# Step 4: Evaluate suitability for K-Means clustering
# Feature scaling
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

# Elbow method to determine optimal number of clusters
inertia = []
k_values = range(1, 11)

for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(scaled_data)
    inertia.append(kmeans.inertia_)

# Plot the elbow curve
plt.figure(figsize=(8, 6))
plt.plot(k_values, inertia, marker='o', linestyle='--')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.grid(True)
plt.show()

# Step 5: Description and evaluation
print("\nDescription and Evaluation:")
print("""
The Banknote Authentication Dataset consists of two continuous features (V1 and V2) with 1372 observations.
Statistical analysis reveals that V1 has a mean of 0.4338 and a standard deviation of 2.6354, while V2 has a mean of 3.7617 and a standard deviation of 4.4607.
Visualizations show distinct groupings in the data, indicating natural clusters. The features exhibit moderate correlation (0.35), suggesting complementary information.

The dataset is suitable for K-Means clustering due to its natural groupings and separable clusters. However, preprocessing steps such as feature scaling are recommended to account for differing scales. Additionally, care should be taken to handle potential outliers that could distort cluster centroids.
""")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from matplotlib.backends.backend_pdf import PdfPages
import os

# Step 1: Load the dataset
data = pd.read_csv('Banknote-authentication-dataset-.csv')

# Feature scaling
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

# Create a folder to save the visualizations
output_dir = "visualization"
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

# Function to save visualizations as PNG and return the file paths
def visualize_and_save(data, scaled_data, output_dir):
    image_paths = []

    # Scatter plot of V1 vs V2
    plt.figure(figsize=(8, 6))
    plt.scatter(data[:, 0], data[:, 1], alpha=0.7)
    plt.title('Scatter Plot of V1 vs V2')
    plt.xlabel('V1')
    plt.ylabel('V2')
    scatter_path = os.path.join(output_dir, 'scatter_plot_v1_v2.png')
    plt.savefig(scatter_path)  # Save as PNG
    plt.close()
    image_paths.append(scatter_path)

    # Histograms for V1 and V2
    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.hist(data[:, 0], bins=30, color='blue', alpha=0.7)
    plt.title('Histogram of V1')
    plt.xlabel('V1')
    plt.ylabel('Frequency')

    plt.subplot(1, 2, 2)
    plt.hist(data[:, 1], bins=30, color='green', alpha=0.7)
    plt.title('Histogram of V2')
    plt.xlabel('V2')
    plt.ylabel('Frequency')

    histograms_path = os.path.join(output_dir, 'histograms_v1_v2.png')
    plt.savefig(histograms_path)  # Save as PNG
    plt.close()
    image_paths.append(histograms_path)

    # Pairwise correlation heatmap
    plt.figure(figsize=(6, 6))
    correlation_matrix = pd.DataFrame(data).corr()
    sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
    plt.title('Correlation Heatmap')
    heatmap_path = os.path.join(output_dir, 'correlation_heatmap.png')
    plt.savefig(heatmap_path)  # Save as PNG
    plt.close()
    image_paths.append(heatmap_path)

    # K-Means clustering visualization
    kmeans = KMeans(n_clusters=2, random_state=42)
    labels = kmeans.fit_predict(scaled_data)
    centroids = kmeans.cluster_centers_

    plt.figure(figsize=(8, 6))
    plt.scatter(scaled_data[:, 0], scaled_data[:, 1], c=labels, cmap='viridis', s=50, alpha=0.7)
    plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='x', s=200, label='Centroids')
    plt.title('K-Means Clustering')
    plt.xlabel('V1 (Scaled)')
    plt.ylabel('V2 (Scaled)')
    plt.legend()
    kmeans_path = os.path.join(output_dir, 'kmeans_clustering.png')
    plt.savefig(kmeans_path)  # Save as PNG
    plt.close()
    image_paths.append(kmeans_path)

    return image_paths

# Generate visualizations and save as PNG
image_paths = visualize_and_save(data.values, scaled_data, output_dir)

# Combine all PNG files into a single PDF
pdf_path = os.path.join(output_dir, 'combined_visualization.pdf')
with PdfPages(pdf_path) as pdf:
    for image_path in image_paths:
        plt.figure(figsize=(8, 6))
        img = plt.imread(image_path)
        plt.imshow(img)
        plt.axis('off')  # Turn off axis
        pdf.savefig()  # Save the current figure to PDF
        plt.close()

print(f"\nAll visualizations have been saved as PNG files in the '{output_dir}' folder.")
print(f"A combined PDF of all visualizations has been saved as '{pdf_path}'.")
```
